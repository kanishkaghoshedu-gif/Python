```python
# #Write a Python program that accepts a sentence and performs the following operations:

# 1. Split the sentence into words and convert each word to lowercase using map() and lambda.
sentence = input("Enter a sentance: ")
words = sentence.split(" ")
lower = list(map(lambda x : x.lower(), words))
print(" ".join(lower))
```

```python
# 2. Filter words having more than 3 characters using filter() and lambda.
sentence = input("Enter a sentance: ")
words = sentence.split(" ")
filtered_words = filter( lambda x: len(x) > 3, words)
print("".join(list(filtered_words)))
```

```python
# 3. Convert the filtered words into uppercase using map() and lambda.
sentence = input("Enter a sentance: ")
words = sentence.split(" ")
filtered_words = filter( lambda x: len(x) > 3, words)

upper = list(map(lambda x : x.upper(), list(filtered_words)))
print(" ".join(upper))
```

```python
# 4. Find the longest word using reduce(). If two words have equal lengths, select the first.
longest_word = reduce(lambda x, y: x if len(x) >= len(y) else y, upper)
print("Longest word:", longest_word)
```

```python
from functools import reduce

sentence = input("Enter a sentence: ")
words = sentence.split(" ")

# 1. Filter words having more than 3 characters
filtered_words = list(filter(lambda x: len(x) > 3, words))

# 2. Map filtered words to uppercase
upper = list(map(lambda x: x.upper(), filtered_words))

# 3. Reduce using '-' separator (with a safety check or initializer if needed)
if upper:
    result = reduce(lambda x, y: x + "-" + y, upper)
    print(result)
else:
    print("No words found with more than 3 characters.")
```

```python
# 7. Calculate the total number of characters in the filtered words using reduce().
from functools import reduce
sentence = input("Enter a sentance: ")
words = sentence.split(" ")
filtered_words = filter( lambda x: len(x) > 3, words)
s=" ".join(list(filtered_words))
print(reduce(lambda x, y: x + len(y), s.split(), 0))
```
