```python
def Divide(a, b):
    div = a/b
    return div


try:
        a = int(input("Enter a: "))
        b = int(input("Enter b: "))
        res = Divide(a, b)
        print(f"a/b = {res}")
except Exception as e:
        print(f"Warning: {e}")
```

```python
def Sum(a, b):
    sum = a+b
    return sum

while(True):
    try:
            a = int(input("Enter a: "))
            b = int(input("Enter b: "))
            res = Sum(a, b)
            print(f"a+b = {res}")
            break;
    except Exception as e:
            print(f"Warning: {e}")
```
